```python
import pandas as pd
```

```python
df1 = pd.DataFrame({
    "ID": [1, 2, 3],
    "Name": ["Ali", "Sara", "Ahmed"]
})


df2 = pd.DataFrame({
    "ID": [1, 2, 3],
    "Marks": [85, 90, 88]
})
```

```python
df2
```

```python
merged_df = pd.merge(df1, df2, on="ID")
```

```python
merged_df
```

```python
df3 = pd.DataFrame({
    "ID": [1, 2, 3],
    "Name": ["Ali", "Sara", "Ahmed"]
})


df4 = pd.DataFrame({
    "ID": [1, 4, 3],
    "Marks": [85, 90, 88]
})

inner_merged_df = pd.merge(df3, df4, on="ID", how="outer")

inner_merged_df
```

```python
df5 = pd.DataFrame({
    "StudentID": [1, 2, 3],
    "Name": ["Ali", "Sara", "Ahmed"]
})


df6 = pd.DataFrame({
    "ID": [1, 2, 3],
    "Marks": [85, 90, 88]
})

result = pd.merge(df5, df6, left_on="StudentID", right_on="ID")

result
```

```python
df1 = pd.DataFrame({
    "Name": ["Ali", "Sara", "Ahmed"]
}, index=[1,2,3])


df2 = pd.DataFrame({
    "Marks": [85, 90, 88]
}, index=[1,2,3])

result = df1.join(df2)
result
```

```python
df3 = pd.DataFrame({
    "ID": [1, 2, 3],
    "Name": ["Ali", "Sara", "Ahmed"]
})


df4 = pd.DataFrame({
    "ID": [1, 4, 3],
    "Marks": [85, 90, 88]
})

df3_index = df3.set_index("ID")
df4_index = df4.set_index("ID")

df3_index.join(df4_index, how="right")
```

```python
df1 = pd.DataFrame({
    "ID": [1, 2, 3],
    "Name": ["Ali", "Sara", "Ahmed"]
})


df2 = pd.DataFrame({
    "ID": [1, 2, 3],
    "Marks": [85, 90, 88]
})

pd.concat([df1, df2],  ignore_index=True)
```
