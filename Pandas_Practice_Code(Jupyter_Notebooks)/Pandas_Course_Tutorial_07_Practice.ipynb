```python
data7 = pd.read_csv("E:\Youtube_Courses\Complete_Pandas_Course\Datasets_Files\grades.csv")

df7 = pd.DataFrame(data7)

df7
```

```python
def grade(marks):
    if marks >= 90:
        return "A"
    elif marks >= 80:
        return "B"
    else:
        return "C"
    
df7["Grade"] = df7["Marks"].apply(grade)

df7
```
